# Registration: thinmme007

Apply the saved transformations to the cell and spot coordinates.

Use Python 3.10 with VALIS 1.1.0, NumPy, pandas and OpenCV, plus the Java and image-reading dependencies required by VALIS. Keep registration_helpers.py beside this notebook.

Inputs: native coordinate tables, the saved registration and its associated files, and the reference coordinate CSVs. Prepared TIFF images are needed only for refitting.


```python
from pathlib import Path
from registration_helpers import get_registrar, read_coordinates, transform_coordinates, check_and_save

data_dir = Path('../../paper_data/Registration_release')
output_dir = Path('outputs/thinmme007/saved_transform_outputs')
refit = False
```

```python
config = {}
config['patient'] = 'thinmme007'
config['registrar'] = 'thinmme007/saved/data/input_registrar.pickle'
config['images'] = 'thinmme007/images'
config['xenium_acquisition'] = 'B-1947702__20260122__XETG00075'
config['xenium_crop_x_um'] = 0
config['visium_crop_xy'] = [25000, 25000]
config['visium_source_sample'] = '702'
```

```python
config['inputs'] = {'xenium': 'thinmme007/native/xenium.csv', 'milan': 'thinmme007/native/milan.csv', 'visium': 'thinmme007/native/visium.csv'}
config['baselines'] = {'xenium': 'thinmme007/baseline/xenium.csv', 'milan': 'thinmme007/baseline/milan.csv', 'visium': 'thinmme007/baseline/visium.csv'}
```

```python
config['fit'] = {'image_type': 'fluorescence', 'max_processed_image_dim_px': 800, 'do_rigid': True, 'compose_non_rigid': True, 'align_to_reference': True, 'check_for_reflections': True, 'max_non_rigid_registration_dim_px': 4500}
config['register'] = {'if_processing_kwargs': {'adaptive_eq': True, 'channel': 'dapi'}}
```

Use the saved transformations for paper reproduction. Seeds are set, but a new VALIS fit can differ between runs and environments.


```python
registrar = get_registrar(config, data_dir, output_dir, refit=refit)
frames = {}
```

Prepare XENIUM coordinates.


```python
native = read_coordinates(config, data_dir, 'xenium')
frames['xenium'] = transform_coordinates(native, 'xenium', config, registrar)
```

Prepare MILAN coordinates.


```python
native = read_coordinates(config, data_dir, 'milan')
frames['milan'] = transform_coordinates(native, 'milan', config, registrar)
```

Prepare VISIUM coordinates.


```python
native = read_coordinates(config, data_dir, 'visium')
frames['visium'] = transform_coordinates(native, 'visium', config, registrar)
```

Compare with the selected saved CSVs. Stop on a mismatch; existing outputs are never replaced.


```python
check_and_save(frames, config, data_dir, output_dir / 'coordinates')
```
